```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator


def load_dataset():
    X_train = np.array([
        [5.1, 3.5, 1.4, 0.2], [4.9, 3.0, 1.4, 0.2], [4.7, 3.2, 1.3, 0.2], [4.6, 3.1, 1.5, 0.2],
        [5.0, 3.6, 1.4, 0.2], [5.4, 3.9, 1.7, 0.4], [4.6, 3.4, 1.4, 0.3], [5.0, 3.4, 1.5, 0.2],
        [4.4, 2.9, 1.4, 0.2], [4.9, 3.1, 1.5, 0.1], [5.4, 3.7, 1.5, 0.2], [4.8, 3.4, 1.6, 0.2],
        [4.8, 3.0, 1.4, 0.1], [4.3, 3.0, 1.1, 0.1], [5.8, 4.0, 1.2, 0.2], [5.7, 4.4, 1.5, 0.4],
        [5.4, 3.9, 1.3, 0.4], [5.1, 3.5, 1.4, 0.3], [5.7, 3.8, 1.7, 0.3], [5.1, 3.8, 1.5, 0.3],
        [5.4, 3.4, 1.7, 0.2], [5.1, 3.7, 1.5, 0.4], [4.6, 3.6, 1.0, 0.2], [5.1, 3.3, 1.7, 0.5],
        [4.8, 3.4, 1.9, 0.2], [5.0, 3.0, 1.6, 0.2], [5.0, 3.4, 1.6, 0.4], [5.2, 3.5, 1.5, 0.2],
        [5.2, 3.4, 1.4, 0.2], [4.7, 3.2, 1.6, 0.2], [4.8, 3.1, 1.6, 0.2], [5.4, 3.4, 1.5, 0.4],
        [5.2, 4.1, 1.5, 0.1], [5.5, 4.2, 1.4, 0.2], [4.9, 3.1, 1.5, 0.1], [5.0, 3.2, 1.2, 0.2],
        [5.5, 3.5, 1.3, 0.2], [4.9, 3.1, 1.5, 0.1], [4.4, 3.0, 1.3, 0.2], [5.1, 3.4, 1.5, 0.2],
        [7.0, 3.2, 4.7, 1.4], [6.4, 3.2, 4.5, 1.5], [6.9, 3.1, 4.9, 1.5], [5.5, 2.3, 4.0, 1.3],
        [6.5, 2.8, 4.6, 1.5], [5.7, 2.8, 4.5, 1.3], [6.3, 3.3, 4.7, 1.6], [4.9, 2.4, 3.3, 1.0],
        [6.6, 2.9, 4.6, 1.3], [5.2, 2.7, 3.9, 1.4], [5.0, 2.0, 3.5, 1.0], [5.9, 3.0, 4.2, 1.5],
        [6.0, 2.2, 4.0, 1.0], [6.1, 2.9, 4.7, 1.4], [5.6, 2.9, 3.6, 1.3], [6.7, 3.1, 4.4, 1.4],
        [5.6, 3.0, 4.5, 1.5], [5.8, 2.7, 4.1, 1.0], [6.2, 2.2, 4.5, 1.5], [5.6, 2.5, 3.9, 1.1],
        [5.9, 3.2, 4.8, 1.8], [6.1, 2.8, 4.0, 1.3], [6.3, 2.5, 4.9, 1.5], [6.1, 2.8, 4.7, 1.2],
        [6.4, 2.9, 4.3, 1.3], [6.6, 3.0, 4.4, 1.4], [6.8, 2.8, 4.8, 1.4], [6.7, 3.0, 5.0, 1.7],
        [6.0, 2.9, 4.5, 1.5], [5.7, 2.6, 3.5, 1.0], [5.5, 2.4, 3.8, 1.1], [5.5, 2.4, 3.7, 1.0],
        [5.8, 2.7, 3.9, 1.2], [6.0, 2.7, 5.1, 1.6], [5.4, 3.0, 4.5, 1.5], [6.0, 3.4, 4.5, 1.6],
        [6.7, 3.1, 4.7, 1.5], [6.3, 2.3, 4.4, 1.3], [5.6, 3.0, 4.1, 1.3], [5.5, 2.5, 4.0, 1.3]
    ])
    y_train = np.array([0] * 40 + [1] * 40)

    X_val = np.array([
        [5.0, 3.5, 1.3, 0.3], [4.5, 2.3, 1.3, 0.3], [4.4, 3.2, 1.3, 0.2], [5.0, 3.5, 1.6, 0.6],
        [5.1, 3.8, 1.9, 0.4], [4.8, 3.0, 1.4, 0.3], [5.1, 3.8, 1.6, 0.2], [4.6, 3.2, 1.4, 0.2],
        [5.3, 3.7, 1.5, 0.2], [5.0, 3.3, 1.4, 0.2], [5.5, 2.6, 4.4, 1.2], [6.1, 3.0, 4.6, 1.4],
        [5.8, 2.6, 4.0, 1.2], [5.0, 2.3, 3.3, 1.0], [5.6, 2.7, 4.2, 1.3], [5.7, 3.0, 4.2, 1.2],
        [5.7, 2.9, 4.2, 1.3], [6.2, 2.9, 4.3, 1.3], [5.1, 2.5, 3.0, 1.1], [5.7, 2.8, 4.1, 1.3]
    ])
    y_val = np.array([0] * 10 + [1] * 10)

    return X_train, y_train, X_val, y_val


def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def forward_step(x, weights, bias):
    z = bias + np.dot(weights, x)
    gz = sigmoid(z)
    pred = 1 if gz >= 0.5 else 0
    return gz, pred


def evaluate(X, y, weights, bias):
    sse, correct = 0.0, 0
    for x_i, target in zip(X, y):
        gz, pred = forward_step(x_i, weights, bias)
        sse += (gz - target) ** 2
        if pred == target:
            correct += 1
    return sse / len(X), correct / len(X)


def train_model(X_train, y_train, X_val, y_val, lr=0.1, epochs=5, bias=0.5, weights=None):
    if weights is None:
        weights = np.array([0.5, 0.5, 0.5, 0.5])

    history = {
        'train_loss': [], 'val_loss': [],
        'train_acc': [], 'val_acc': []
    }

    for _ in range(epochs):
        train_sse, train_correct = 0.0, 0
        
        # Training loop
        for x_i, target in zip(X_train, y_train):
            gz, pred = forward_step(x_i, weights, bias)
            
            err = gz - target
            train_sse += err ** 2
            if pred == target:
                train_correct += 1
                
            # Derivative calculations matching GSheet logic
            dbias = 2 * err * gz * (1 - gz)
            dweights = dbias * x_i
            
            # Parameter updates
            bias -= lr * dbias
            weights -= lr * dweights
            
        history['train_loss'].append(train_sse / len(X_train))
        history['train_acc'].append(train_correct / len(X_train))

        # Validation loop
        val_loss, val_acc = evaluate(X_val, y_val, weights, bias)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)

    return history


def print_metrics(history):
    epochs = len(history['train_acc'])
    for i in range(epochs):
        print(f"Training Accuracy: {history['train_acc'][i]:.4f}, Validation Accuracy: {history['val_acc'][i]:.4f}")
        print(f"Training Loss: {history['train_loss'][i]:.4f}, Validation Loss: {history['val_loss'][i]:.4f}\n")


def plot_metric(epochs, train_data, val_data, title, ylabel, filename):
    plt.figure(figsize=(6, 4))
    plt.plot(range(1, epochs + 1), train_data, label=f'Training {ylabel}', marker='o')
    plt.plot(range(1, epochs + 1), val_data, label=f'Validation {ylabel}', marker='s')
    plt.title(title)
    plt.xlabel('Epoch')
    plt.ylabel(ylabel)
    plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))
    plt.legend()
    plt.grid(True)
    plt.show()


def main():
    X_train, y_train, X_val, y_val = load_dataset()

    epochs = 5
    lr = 0.1

    history = train_model(
        X_train, y_train, X_val, y_val, lr=lr, epochs=epochs
    )

    print_metrics(history)

    plot_metric(epochs, history['train_acc'], history['val_acc'], 'Accuracy Chart (Python)', 'Accuracy', 'accuracy_chart_python.png')
    plot_metric(epochs, history['train_loss'], history['val_loss'], 'Loss Chart (Python)', 'Loss (MSE)', 'loss_chart_python.png')

if __name__ == '__main__':
    main()
```
